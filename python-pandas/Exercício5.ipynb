### Tratamento de Strings e Datas (Limpeza)
#### Tarefa: Crie um DataFrame com uma coluna Nome_Usuario (ex: "  JOÃO silva  ") e uma coluna Data_Abertura (ex: "2023-05-#### 20").
#### Limpe o nome usando .str.strip() e deixe tudo em minúsculo com .lower().
#### Converta a coluna de data para o formato correto usando pd.to_datetime().

```python
import pandas as pd
import numpy as np
```

```python
data = {'Nome_Usuario' : ['   JOÃO silva   ', '   miguel REMIDES   '],
       'Data_Abertura' : ['2023-05-20', '2006-07-01']
       }
df = pd.DataFrame(data)
df
```

```python
df ['Nome_Usuario'] = df['Nome_Usuario'].str.strip().str.lower()
df
```

```python
df['Data_Abertura'] = pd.to_datetime(df['Data_Abertura'])
df
```
