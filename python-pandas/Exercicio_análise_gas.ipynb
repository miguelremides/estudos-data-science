##### 1. Carregue os conjuntos de dados "gasolina_2000+.csv" e "gasolina_2010+.csv" em dois DataFrames diferentes e combine-os em um único DataFrame.

```python
df1 = pd.read_csv('gasolina_2000+.csv', low_memory = False)
df2 = pd.read_csv('gasolina_2010+.csv', low_memory = False)
df3 = pd.concat([df1, df2], ignore_index = True)
df3
```

##### 4. Você deve ter percebido que as colunas DATA INICIAL e DATA FINAL estão formatadas como string. Utilizando o método pd.to_datetime, converta ambas as colunas para Timestamp / Datetime.

```python
df3['DATA INICIAL'] = pd.to_datetime(df3['DATA INICIAL'])
df3['DATA FINAL'] = pd.to_datetime(df3['DATA FINAL'])
df3[['DATA INICIAL', 'DATA FINAL']].info()
```

##### 5.Crie uma nova coluna para representar o mês e o ano (mm-aaaa), utilizando a coluna DATA FINAL como referência.

```python
df3['MÊS-ANO'] = df3['DATA FINAL'].dt.strftime('%m-%Y')
df3.head()
```

##### 6.Utilizando o value_counts(), liste todos os tipos de produtos contidos na base de dados.

```python
df3['PRODUTO'].value_counts()
```

##### 7. Filtre o DataFrame para obter apenas dados da 'GASOLINA COMUM'. Grave em uma nova variável.

```python
df_gasolina_comum = df3[df3['PRODUTO'] == 'GASOLINA COMUM']
df_gasolina_comum
```

##### 8. Qual o preço médio de revenda da gasolina em agosto de 2008?

```python
df3.head()
```

```python
dados_agosto_2008 = df_gasolina_comum[df_gasolina_comum['MÊS-ANO'] == '08-2008']
preço_medio_revenda = dados_agosto_2008['PREÇO MÉDIO REVENDA'].mean()
print(f'O preço médio de revenda em agosto de 2008 foi {preço_medio_revenda:.1f}R$')
```

##### 9. Qual o preço médio de revenda da gasolina em maio de 2014 em São Paulo?

```python
dados_maio_2014 = df_gasolina_comum[df_gasolina_comum['MÊS-ANO'] == '05-2014' ]
dados_maio_2014_sp = dados_maio_2014[dados_maio_2014['ESTADO'] == 'SAO PAULO']
preço_medio2 = dados_maio_2014_sp['PREÇO MÉDIO REVENDA'].mean()
print(f'O preço médio de revenda da gasolina em maio de 2014 em São Paulo foi de {preço_medio2:.1f}R$')
```

##### 10. Você conseguiria descobrir em qual(quais) estado(s) a gasolina ultrapassou a barreira dos 5,00 R$? E quando isso occorreu?

```python
df_gasolina_maior_5 = df_gasolina_comum[df_gasolina_comum['PREÇO MÉDIO REVENDA'] > 5]
df_gasolina_maior_5['ESTADO'].value_counts()
```

##### 11. Qual a média de preços dos estados da região sul em 2012?

```python
df_sul = df3[df3['REGIÃO'] == 'SUL']
df_sul_gasolina = df_sul[df_sul['PRODUTO'] == 'GASOLINA COMUM']
df_sul_gasolina_2012 = df_sul_gasolina[df_sul_gasolina['DATA FINAL'].dt.year == 2012]
preço_medio_sul = df_sul_gasolina_2012['PREÇO MÉDIO REVENDA'].mean()
print(f'O preço médio da gasolina nos estados da região Sul em 2012 é igual a {preço_medio_sul:.1f}R$.')
```

##### 12. Você conseguiria obter uma tabela contendo a variação percentual ano a ano para o estado do Rio de Janeiro?

```python
df_rj = df_gasolina_comum[df_gasolina_comum['ESTADO'] == 'RIO DE JANEIRO']
precos_anuais = df_rj.groupby(df_rj['DATA FINAL'].dt.year)['PREÇO MÉDIO REVENDA'].mean()
variacao_percentual = precos_anuais.pct_change() * 100
tabela_variacao = (variacao_percentual.to_frame(name = 'VARIAÇÃO %')).style.format("{:.2f}%")
tabela_variacao
```

##### Desafio: Crie uma tabela contendo a séria temporal com a diferença absoluta e percentual entre os valores mais baratos e caros. Apresente também ao lado os estados na qual os maiores e os menores preços foram registrados.

```python
idx_min = df_gasolina_comum.groupby('MÊS-ANO')['PREÇO MÉDIO REVENDA'].idxmin()
idx_max = df_gasolina_comum.groupby('MÊS-ANO')['PREÇO MÉDIO REVENDA'].idxmax()

df_desafio = pd.DataFrame()
df_desafio['Mês'] = df_gasolina_comum.loc[idx_min, 'MÊS-ANO'].values
df_desafio['Preço Mínimo'] = df_gasolina_comum.loc[idx_min, 'PREÇO MÉDIO REVENDA'].values
df_desafio['Estado (Mín)'] = df_gasolina_comum.loc[idx_min, 'ESTADO'].values
df_desafio['Preço Máximo'] = df_gasolina_comum.loc[idx_max, 'PREÇO MÉDIO REVENDA'].values
df_desafio['Estado (Máx)'] = df_gasolina_comum.loc[idx_max, 'ESTADO'].values

df_desafio['Diferença Absoluta'] = df_desafio['Preço Máximo'] - df_desafio['Preço Mínimo']
df_desafio['Diferença %'] = (df_desafio['Diferença Absoluta'] / df_desafio['Preço Mínimo']) * 100

display(df_desafio.head(10))
```
