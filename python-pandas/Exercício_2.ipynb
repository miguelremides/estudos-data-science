# DataFrames e Filtros de Crédito

##### Objetivo: Simular a estrutura do seu projeto de risco de crédito usando DataFrames e seleções condicionais.
##### Crie um DataFrame com 5 linhas e as colunas: Idade, Renda, Empréstimo e Inadimplente (True/False).
##### Crie uma nova coluna chamada Comprometimento que seja o resultado da divisão da coluna Empréstimo pela Renda.
##### Utilize o método .loc para selecionar apenas as linhas onde a Idade é superior a 25 anos.
##### Filtre o DataFrame para exibir apenas os clientes que são Inadimplente == True E possuem Renda > 3000.

```python
import pandas as pd
import numpy as np
```

```python
dados = {'Idade': [54, 28, 19, 70, 39], 
         'Renda': [5000, 2500, 1600, 8000, 4500], 
         'Empréstimo': [1000, 500, 2500, 300, 20000], 
         'Inadimplente': [True, False, False, True, False]} 
```

```python
df = pd.DataFrame(dados)
df
```

```python
df['Comprometimento'] = df['Empréstimo']/df['Renda']
df
```

```python
df.loc[df['Idade'] > 25]
```

```python
df[(df['Inadimplente']) & (df['Renda'] > 3000)]
```
