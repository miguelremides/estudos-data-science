# Agrupamento de Vendas (BI Style)

#### Objetivo: Aplicar o Groupby para gerar insights de negócio, como em uma vaga de BI.
#### Use o dataset abaixo para o exercício:

```python
import pandas as pd
data = {
    'Regiao': ['Norte', 'Norte', 'Sul', 'Sul', 'Centro', 'Centro'],
    'Vendedor': ['Ana', 'Bruno', 'Caio', 'Duda', 'Elias', 'Fernanda'],
    'Vendas': [2500, 1800, 3200, 4100, 1500, 2900],
    'Metas_Batidas': [1, 0, 1, 1, 0, 1]
}
df = pd.DataFrame(data)
```

#### Agrupe os dados por Regiao.
#### Calcule a soma total de vendas por região, garantindo que apenas colunas numéricas sejam somadas (numeric_only=True).
#### Calcule a média de vendas por região.
#### Insight: Qual região teve o maior valor individual de venda?

```python
df
```

```python
df.groupby('Regiao')
```

```python
df.groupby('Regiao')[['Vendas']].sum(numeric_only = True)
```

```python
df.groupby('Regiao')[['Vendas']].mean(numeric_only = True)
```
