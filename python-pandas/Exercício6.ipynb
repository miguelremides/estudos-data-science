### Análise de Risco.
#### No seu DataFrame, crie as colunas Valor_Emprestimo (ex: 10000, 5000) e Numero_Parcelas
#### Crie a coluna Comprometimento_Mensal: divida o valor do empréstimo pelo número de parcelas.
#### Filtro: Mostre apenas as linhas onde o Comprometimento_Mensal é maior que 500.

```python
import pandas as pd
import numpy as np
```

```python
data = {'Nome_Usuario' : ['   JOÃO silva   ', '   miguel REMIDES   '],
       'Data_Abertura' : ['2023-05-20', '2006-07-01']
       }
df = pd.DataFrame(data)
df ['Nome_Usuario'] = df['Nome_Usuario'].str.strip().str.lower()
df['Data_Abertura'] = pd.to_datetime(df['Data_Abertura'])
df
```

```python
df['Valor_Emprestimo'] = [20000, 5000]
df['Numero_Parcelas'] = [24, 12]
df
```

```python
df['Comprometimento_Mensal'] =  df['Valor_Emprestimo']/df['Numero_Parcelas']
df
```

```python
df[df['Comprometimento_Mensal'] > 500]
```
