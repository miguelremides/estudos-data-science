## O "JOIN" do Pandas (Merge)
#### Tarefa: Crie dois DataFrames: df_perfil (com colunas ID_Cliente, Nome, Renda) e df_score (com ID_Cliente, Pontuacao_Credito).
#### Ação: Use o pd.merge() para unir as duas tabelas pelo ID_Cliente. Garanta que mesmo o cliente que não tem score apareça na lista final.

```python
import pandas as pd
import numpy as np
```

```python
perfil = {
    'ID_Cliente': [101, 102, 103, 104],
    'Nome': ['Miguel Remides', 'Anna Mushchenko', 'Soraya Viviane', 'Vinicius Remides'],
    'Renda': [5000, 4500, 6000, 2000]
}

score = {
    'ID_Cliente' : [101, 102, 103], 
    'Pontuacao_Credito': [10, 9, 4]
}
df_perfil = pd.DataFrame(perfil)
df_score = pd.DataFrame(score)
```

```python
df_perfil
```

```python
df_score
```

```python
pd.merge(df_perfil, df_score, on = 'ID_Cliente')
```

```python
df_final = pd.merge(df_perfil, df_score, on='ID_Cliente', how='left')
df_final
```

```python
df_final['Pontuacao_Credito'] = df_final['Pontuacao_Credito'].fillna(0)
df_final
```
